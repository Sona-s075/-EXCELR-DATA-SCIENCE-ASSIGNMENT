## Confidence Intervals

```python
import numpy as np
import scipy.stats as stats
import pandas as pd
# Sample data
data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)  # ddof=1 for sample standard deviation
sample_size = len(data)
sample_mean = np.mean(data)
sample_std_dev = np.std(data)
n = len(data)
df1 = n-1
print("Sample Mean:",sample_mean)
print("Sample Standard deviation is:",sample_std_dev)
```

Section-A:Build 99% Confidence Interval Using Sample Standard Deviation

```python
from scipy.stats import norm
#alpha = 1%
ci_99 = norm.interval(0.99,loc=np.mean(data),scale = np.std(data))
print("99% confidence interval for the mean number of characters printed before the print-head fails using the sample standard deviation:",ci_99)
```

T - Distribution:

T - distributoin can be used only when the sample data size is less than 30.

```python
#t distribution for 99% confidence interval
t_value = stats.t.ppf(0.995,df1)
t_value
```

```python
error_sample = t_value*(sample_std_dev / np.sqrt(n))
error_sample
```

Section-B:. Build 99% Confidence Interval Using Known Population Standard Deviation

```python
population_std_dev = 0.2
```

```python
#z-value for 99% confidence interval
z_value = stats.norm.ppf(0.995)
z_value
```

```python
#Error using known population standard deviation
error_population = z_value*(population_std_dev / np.sqrt(n))
error_population
```

```python
#alpha value 1%
ci99 = norm.interval(0.99,loc = np.mean(data),scale=population_std_dev)
ci99
print("construct a 99% confidence interval for the mean number of characters printed before failure:",ci99)
```

## Hypothesis Testing

```python
import numpy as np
import math
from scipy.stats import norm
# Step 1: State the Hypotheses

# H0: The weekly operating cost follows the theoretical model (mean = 4000 Rs)
# H1: The weekly operating cost is higher than the theoretical model (mean > 4000 Rs)

# Step 2: Calculate the Test Statistic

sample_mean = 3050   # observed mean weekly cost
X_mean = 600         # mean units produced
X_std = 25           # std deviation of units produced
n = 25               # sample size
# Theoretical mean weekly cost according to model
mu = 1000 + 5 * X_mean   # W = 1000 + 5X
print("Theoretical mean weekly cost:", mu)

# Standard deviation of weekly cost
sigma = 5 * X_std
print("Standard deviation of weekly cost:", sigma)

# Standard error
SE = sigma / math.sqrt(n)

# Test statistic (z)
z = (sample_mean - mu) / SE
print("Test Statistic (z):", z)

# Step 3: Determine Probability
alpha = 0.05
critical_value = norm.ppf(1 - alpha)  # one-tailed test
p_value = 1 - norm.cdf(z)
print("Critical Value (z_critical):", critical_value)
print("p-value:", p_value)

# Step 4: Make a Decision
if z > critical_value:
    decision = "Reject H0"
else:
    decision = "Fail to Reject H0"

print("Decision:", decision)
# Step 5: Conclusion
if decision == "Reject H0":
    conclusion = ("There is strong evidence to support the restaurant owners' claim "
                  "that the weekly operating costs are higher than the model suggests.")
else:
    conclusion = ("There is not enough evidence to support the claim. "
                  "The weekly operating costs do not appear significantly higher than the model suggests.")

print("Conclusion:", conclusion)
```
